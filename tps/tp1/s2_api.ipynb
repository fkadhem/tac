## Utiliser [Wikipedia](https://www.mediawiki.org/wiki/API:Action_API) pour connaître la définition d'un mot

```python
"""Retrieve definition from Wikipedia"""
headers = {'User-Agent': 'https://github.com/fkadhem'}

URL = "https://fr.wikipedia.org/w/api.php"

S = requests.Session()

search_term = "Prusse" # présent dans le corpus Camille KB_JB92_1860-02-09_01-00003

PARAMS = {
    "action": "query",
    "format": "json",
    "list": "search",
    "srsearch": search_term
}
R = S.get(url=URL, params=PARAMS, headers=headers)
DATA = R.json()

if DATA['query']['search'][0]['title'] == search_term:
    lien = f"https://fr.wikipedia.org/wiki/{search_term}"
    print("Votre recherche " + search_term + " existe sur Wikipedia ")
    print("Lien :" + lien)
```
