# Reconnaissance d'entités nommées avec SpaCy

La documentation est accessible ici: https://spacy.io/api

## Imports

```python
from collections import defaultdict
import spacy
from spacy.lang.fr.examples import sentences
```

```python
nlp = spacy.load('fr_core_news_md')
```

## Appliquer la reconnaissance d'entités nommées sur notre corpus

```python
# Charger le texte
n=1000000
text = open("../../data/tmp/1969_clean.txt", encoding='utf-8').read()[:n]
```

```python
%%time
# Traiter le texte

doc = nlp(text)
```

```python
# Compter les entités
entities = defaultdict(int)
for ent in doc.ents:
    if (ent.label_ == "PER" or ent.label_ == "ORG" or ent.label_ == "LOC") and len(ent.text) > 3:
        entities[f"{ent.text}({ent.label_})"] += 1
```

```python
# Trier et imprimer

sorted_entities = sorted(entities.items(), key=lambda kv: kv[1], reverse=True)

for ent, freq in sorted_entities:
    print(f"{ent} apparait {freq} fois dans le corpus")
```
